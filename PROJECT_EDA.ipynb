2. Interquartile Range (IQR) method :

* IQR method is a statistical technique used primarily to measure variability and detect outliers in a dataset.

IQR = Q_3 - Q_1

Q_1 (First Quartile): The median of the lower half of the dataset (the middle number between the smallest number and the median).

Q_2 (Second Quartile): The median of the entire dataset.

Q_3 (Third Quartile): The median of the upper half of the dataset (the middle value between the median and the highest value).

```python
def iqr_outlier_summary(data, numerical_cols):
  summary_list = []

  for col in numerical_cols:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Filter outliers
    outliers = data[(data[col] < lower_bound) | (data[col] > upper_bound)]
    outlier_count = len(outliers)
    outlier_percentage = (outlier_count / len(data)) * 100

    summary_list.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage (%)": round(outlier_percentage, 2),
    })

  return pd.DataFrame(summary_list)


# Running the function in numerical columns
iqr_summary_df = iqr_outlier_summary(EDA, numerical_columns)
display(iqr_summary_df)
```

Multivariate analysis: Examines the simultaneous relationships among three or more variables to uncover complex patterns, interactions, and dependencies in a dataset.

```python
# Faceted or Multi-variable Scatter Plot: Discount vs Final Price grouped by Product Category
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=EDA,
    x="discount_percentage",
    y="final_price",
    hue="product_category",
    size="quantity",
    palette="Set2",
    alpha=0.7,
    sizes=(20, 150)
)
plt.title("Discount Percentage vs Final Price by Product Category & Quantity")
plt.xlabel("Discount Percentage (%)")
plt.ylabel("Final Price")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Multi-panel multivariate bubble plot
g = sns.relplot(
    data=EDA,
    x="product_price",
    y="final_price",
    hue="product_category",
    size="discount_percentage",
    col="return_status",
    row="shipping_type",
    palette="Set2",
    alpha=0.7,
    height=4,
    aspect=1.5,
    sizes=(20, 200)
)

g.fig.suptitle("Multivariate Analysis: Product Category × Discount × Final Price", y=1.03, fontsize=14, fontweight="bold")
g.set_axis_labels("Product Price (USD)", "Final Price (USD)")
g.tight_layout()
plt.show()
```

Skewness and Kurtosis Interpretation:

We analyzed the distribution shape of our numerical columns (`age`, `product_price`, `quantity`, `discount_percentage`, `final_price`, and `delivery_days`) using skewness and kurtosis metrics:

### 1. Skewness
* **Positive Skew in `final_price` (0.935):** Unlike other features which are symmetrically or uniformly distributed, the final transaction value is right-skewed. This indicates that the majority of transactions are small-to-medium-sized, with a long tail of high-value purchases.
* **Near-Zero Skew:** Variables like `age`, `product_price`, `quantity`, and `discount_percentage` show skewness values close to 0, implying well-balanced or uniform distributions (e.g., discounts evenly spread between 0-30%).

### 2. Kurtosis
* **High Kurtosis in `final_price` (0.169 vs negative kurtosis in others):** The heavy tail in final revenue highlights the presence of extreme high-value transactions (driven by bulk quantities or premium items). Other variables show platykurtic distributions (flat-topped with thin tails).
* **Business Takeaway:** Revenue is largely concentrated on limited high-value purchases rather than uniform spending, which validates keeping these outliers as genuine premium consumer behaviors rather than deleting them.

```python
# Detailed Skewness and Kurtosis Summary Table
distribution_summary = pd.DataFrame(
    {
        "Variable": numerical_columns,
        "Skewness": EDA[numerical_columns].skew().values,
        "Kurtosis": EDA[numerical_columns].kurtosis().values,
        "Distribution Type": [
            "Symmetric / Normal",
            "Uniform / Symmetric",
            "Uniform",
            "Uniform",
            "Right-Skewed (Positive)",
            "Symmetric",
        ],
    }
).round(3)

display(distribution_summary)
```
